### What are keyword arguments? Give an example.

```python

def addition(**kwarg):
    """
    Keyword Arguments- keyword arguments defined when we don't know that how variable and values arguments are going to be passed at the time of 
    calling function. So there we use keyword arguments. It gives values as a dictionary to function.it gives variable name as a key and value 
    as a value in dictionary.

    """
    return sum(kwarg.values())

print("Addition= ",addition(a=12,b=125,c=45,d=47))
print(addition.__doc__)
```

### What does **kwargs mean? Write a function that prints all keyword arguments using **kwargs.

```python
def example(**kwargs):
    """
    **kwargs is used when we don't know that how many number of variable and values are going to be passed at the time of function
    calling, So at that time we use it. it give variable and value as a dictionary to the function.
    """
    for i,j in kwargs.items():
        print(i,j)

example(a=120,b=124,c=45,d=546,e=76,f=45)
print(example.__doc__)
```

### Write an adapter function that takes a list of operations and executes add, sub, mul, div based on input. 

### Example: [ {"op": "add", "a": 10, "b": 20}, {"op": "mul", "a": 4, "b": 5} ]

```python
def adapter(ls):
    for i in ls:
        if i["op"]=="add":
            print(f"Addition of {i["a"]} and {i["b"]} = ",i["a"]+i["b"])
        
        elif i["op"]=="sub":
            print(f"Subtraction of {i["a"]} and {i["b"]} = ",i["a"]-i["b"])
        
        elif i["op"]=="mul":
            print(f"Multiplication of {i["a"]} and {i["b"]} = ",i["a"]*i["b"])
        
        else:
            print(f"Division of {i["a"]} and {i["b"]} = ",i["a"]//i["b"])
        

ls=[ {"op": "add", "a": 10, "b": 20}, {"op": "mul", "a": 4, "b": 5},{"op": "sub", "a": 50, "b": 37},{"op": "div", "a": 45, "b": 5} ]
adapter(ls)

```

### Write a function process_students(data) that prints total marks and grade for each student.
- Example: [ {"name":"A", "marks":[80,90,70]}, {"name":"B", "marks":[60,75,65]} ] 
- Expected Output: Name: A | Total: 240 | Grade: A Name: B | Total: 200 | Grade: B

```python
def process_students(data):
    for i in data:
        if sum(i['marks'])/3>85:
            grade='A'
        elif 85>sum(i['marks'])/3>70:
            grade='B'
        elif 70>sum(i['marks'])/3>40:
            grade='C'
        elif sum(i['marks'])/3<40:
            grade='F'

        print(f"Name: {i['name']} | Total: {sum(i['marks'])} | Grade: {grade}")
    

data=[ {"name":"A", "marks":[80,90,70]}, {"name":"B", "marks":[60,75,65]} ]
process_students(data)
```

### Create a function apply_discounts(products) that updates price after discount. 
- Example: [ {"name":"Laptop", "price":50000, "discount":10}, {"name":"Phone", "price":20000, "discount":5} ] 
- Expected Output: Laptop final: 45000 Phone final: 19000

```python
def apply_discounts(products):
    for i in products:
        dispr=i['price']-i['price']*(i['discount']/100)
        print(f"{i['name']} final: {dispr}")

data=[ {"name":"Laptop", "price":50000, "discount":10}, {"name":"Phone", "price":20000, "discount":5} ] 
apply_discounts(data)
```

### Write a function generate_report(employees) that calculates bonus based on salary slabs.

### Write a function average_salary(emp_list) that returns the average salary.

### Write a function salary_hike(emp_list, percent) that increases each salary by given percent.

```python
def salary_hike(emp_list,percent):
    print("Before Salary hike-")
    for i in emp_list:
        print(f"ID: {i['id']}   Name: {i['name']}       Age: {i['age']}      Dept.:  {i['department']}     Salary: {i['salary']}")
    print(f"\nAfter {percent}% Salary hike")
    for i in emp_list:
        i['salary']+=i['salary']*(percent/100)
        print(f"ID: {i['id']}   Name: {i['name']}       Age: {i['age']}      Dept.:  {i['department']}     Salary: {i['salary']}")

data=[ {"id": 101, "name": "Amit", "age": 28, "department": "IT", "salary": 55000}, 
      {"id": 102, "name": "Neha", "age": 24, "department": "HR", "salary": 40000}, 
      {"id": 103, "name": "Raj", "age": 31, "department": "Finance", "salary": 65000}, 
      {"id": 104, "name": "Simran", "age": 29, "department": "IT", "salary": 60000}, 
      {"id": 105, "name": "Vijay", "age": 35, "department": "Admin", "salary": 35000} ]

salary_hike(data,15)
```

### Write out_of_stock(items) that returns all products with stock < 20.

```python
def out_of_stock(items):
    oos=[]
    for i in items:
        if i['stock']<20:
            oos.append(i)
    return oos

products = [
    {'id': 1, 'name': 'Laptop', 'stock': 15},
    {'id': 2, 'name': 'Mouse', 'stock': 45},
    {'id': 3, 'name': 'Keyboard', 'stock': 19},
    {'id': 4, 'name': 'Monitor', 'stock': 25},
    {'id': 5, 'name': 'Webcam', 'stock': 5}
]

lsoos=out_of_stock(products)
print("Out of Stock items:  ")
for i in lsoos:
    print(f"ID:{i['id']}    Name:{i['name']}    Stock:{i['stock']}")
```

### Write search_product(items, name) → return matching product(s).
